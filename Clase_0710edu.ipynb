```python
#Bloque C - Ejercicio 9
mensaje = input("Escribe un mensaje: ").lower()

if "hola" in mensaje or "buenas" in mensaje:
    print("Saludo detectado")
else:
    print("Sin saludo")
```

```python
#Ejercicio 10 – Clasificador de tickets
mensaje = input("Escribe tu consulta: ").lower()

if "problema" in mensaje or "error" in mensaje or "ayuda" in mensaje:
    print("Soporte")
elif "precio" in mensaje or "comprar" in mensaje or "venta" in mensaje:
    print("Ventas")
else:
    print("Otro")
```

```python
#Ejercicio 11 – Contar palabras sospechosas
mensaje = input("Escribe un mensaje: ").lower()
palabras = ["premio", "gratis", "ganaste", "urgente"]
contador = 0

for palabra in palabras:
    if palabra in mensaje:
        contador += 1

print("Palabras sospechosas encontradas:", contador)
```

```python
#Ejercicio 12 – Sentimiento básico
mensaje = input("Escribe un mensaje: ").lower()

positivas = ["feliz", "bueno", "excelente", "genial"]
negativas = ["malo", "triste", "horrible", "terrible"]

puntaje = 0

for palabra in positivas:
    if palabra in mensaje:
        puntaje += 1

for palabra in negativas:
    if palabra in mensaje:
        puntaje -= 1

if puntaje > 0:
    print("Sentimiento positivo")
elif puntaje < 0:
    print("Sentimiento negativo")
else:
    print("Sentimiento neutral")
```

```python
#Ejercicio 13 – Repetir hasta salir
mensaje = ""

while mensaje != "salir":
    mensaje = input("Escribe un mensaje o 'salir' para terminar: ").lower()

    if mensaje != "salir":
        print("Mensaje recibido:", mensaje)

print("Programa terminado")
```

```python
#Ejercicio 14 – Palíndromo
palabra = input("Escribe una palabra: ").lower()

if palabra == palabra[::-1]:
    print("Es un palíndromo")
else:
    print("No es un palíndromo")
```
