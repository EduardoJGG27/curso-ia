```python
#Bloque C - Ejercicio 9
#Solicite un mensaje.
#Si contiene “hola” o “buenas”, mostrar “Saludo detectado”; en caso contrario, “Sin saludo”.

print("Ingrese un mensaje: ")
mensaje = input().lower()

if "hola" in mensaje or "buenas" in mensaje:
    print("Saludo detectado")
else:
    print("Sin saludo")
```

```python
#Ejercicio 10 – Clasificador de tickets
#Clasifique un mensaje como Soporte, Ventas u Otro utilizando palabras clave definidas por usted.

print("Cual es el motivo de su consulta?? ")
mensaje = input().lower()

if "ayuda" in mensaje:
    print("Soporte")
elif ("vender" in mensaje) or ("precio" in mensaje) or ("dinero" in mensaje):
    print("Ventas")
else:
    print("Otro")
```

```python
#Ejercicio 11 – Contar palabras sospechosas
#Recorra una lista de palabras sospechosas y cuente cuántas aparecen en el mensaje.

mensaje = input("Ingrese un mensaje: ").lower()

palabras = ["premio", "gratis", "urgente", "dinero", "5km", "iphone"]

contador = 0

for palabra in palabras:
    if palabra in mensaje:
        contador += 1

print("Cantidad de palabras sospechosas:", contador)
```

```python
#Ejercicio 12 – Sentimiento básico
#Defina palabras positivas y negativas. 
#Recorra ambas listas, calcule un puntaje y clasifique el texto como positivo, neutral o negativo.
mensaje = input("Ingrese un mensaje: ").lower()

positivas = ["feliz", "contento", "excelente", "bueno"]
negativas = ["triste", "enojado", "malo", "cansado"]

puntaje = 0

for palabra in positivas:
    if palabra in mensaje:
        puntaje += 1

for palabra in negativas:
    if palabra in mensaje:
        puntaje -= 1

if puntaje > 0:
    print("Positivo")
elif puntaje < 0:
    print("Negativo")
else:
    print("Neutral")
```

```python
#Ejercicio 13 – Repetir hasta salir
#Use while para solicitar mensajes repetidamente hasta que el usuario escriba “salir”.
mensaje = ""

while mensaje != "salir":
    mensaje = input("Ingrese un mensaje: ").lower()

print("Programa finalizado")
```

```python
#Ejercicio 14 – Palíndromo
#Solicite una palabra y determine si se lee igual de izquierda a derecha y de derecha a izquierda.
palabra = input("Ingrese una palabra: ").lower()

invertida = ""

for letra in palabra:
    invertida = letra + invertida

if palabra == invertida:
    print("Es un palíndromo")
else:
    print("No es un palíndromo")
```
